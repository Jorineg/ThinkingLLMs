```python
import torch
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="google/gemma-2-2b-it",
    model_kwargs={"torch_dtype": torch.bfloat16},
    device="cuda",  # replace with "mps" to run on a Mac device
)

messages = [
    {"role": "user", "content": "Who are you? Please, answer in pirate-speak."},
]

outputs = pipe(messages, max_new_tokens=256)
assistant_response = outputs[0]["generated_text"][-1]["content"].strip()
print(assistant_response)
# Ahoy, matey! I be Gemma, a digital scallywag, a language-slingin' parrot of the digital seas. I be here to help ye with yer wordy woes, answer yer questions, and spin ye yarns of the digital world.  So, what be yer pleasure, eh? 🦜
```

```python
import dns.resolver

def check_dns_records(domain):
    record_types = ['A', 'CNAME', 'MX']  # DNS record types to check
    results = {}
    
    for record_type in record_types:
        try:
            answers = dns.resolver.resolve(domain, record_type)
            records = [str(answer) for answer in answers]
            results[record_type] = records
        except dns.resolver.NoAnswer:
            results[record_type] = "No answer"
        except dns.resolver.NXDOMAIN:
            results[record_type] = "No such domain"
            break  # If NXDOMAIN, no need to check other records
        except dns.resolver.Timeout:
            results[record_type] = "Timed out"
        except dns.resolver.NoNameservers:
            results[record_type] = "No nameservers available"
    
    return results

# List of subdomains to check
subdomains = ['staging.anthropic.com', 'console-staging.anthropic.com', "claude-staging.anthropic.com", "d2vibd1uzjg2oc.cloudfront.net"]


for subdomain in subdomains:
    print(f"\nChecking DNS records for {subdomain}:")
    dns_results = check_dns_records(subdomain)
    for record_type, result in dns_results.items():
        print(f"  {record_type} Record: {result}")
```
